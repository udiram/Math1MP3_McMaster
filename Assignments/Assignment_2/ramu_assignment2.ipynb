```python
# Question 2

def spell_number(n):
    """ spell a given integer between 1 and 9999
    """
        # Dictionaries for number words
    ones = ["", "one", "two", "three", "four", "five", "six", "seven", "eight", "nine"]
    teens = ["", "eleven", "twelve", "thirteen", "fourteen", "fifteen", "sixteen",
             "seventeen", "eighteen", "nineteen"]
    tens = ["", "ten", "twenty", "thirty", "forty", "fifty", "sixty", "seventy", "eighty", "ninety"]
    thousands = ["", "thousand"]

    def spell_hundreds(num):
        if num == 0:
            return ""
        result = ""
        if num >= 100:
            result += ones[num // 100] + " hundred"
            num %= 100
            if num > 0:
                result += " "
        if 10 < num < 20:  # Handle teens
            result += teens[num - 10]
        else:
            if num >= 10:
                result += tens[num // 10]
                num %= 10
                if num > 0:
                    result += " "
            if num > 0:
                result += ones[num]
        return result

    if not (1 <= n <= 9999):
        return "Number out of range"

    # Handle thousands
    result = ""
    if n >= 1000:
        result += ones[n // 1000] + " thousand"
        n %= 1000
        if n > 0:
            result += " "

    # Handle the rest under 1000
    result += spell_hundreds(n)
    return result
    
    
    
print(spell_number(9307)) # should print nine thousand three hundred seven
print(spell_number(5004)) # should print five thousand four
print(spell_number(1616)) # should print one thousand six hundred sixteen
print(spell_number(13)) # should print thirteen
print(spell_number(906)) # should print nine hundred six
print(spell_number(67)) # should print sixty seven
print(spell_number(632)) # should print six hundred thirty two
print(spell_number(2)) # should print two
print(spell_number(111)) # should print one hundred eleven
```

```python
# Question 5

### run this cell to import data
### NOTE: YOU MUST HAVE UPLOADED THE DATA FILE TO JUPYTER IN ORDER TO READ THE DATA ###

import pandas as pd 
data=pd.read_csv("concentrationdata.csv") 

time=[]
conc=[]

for i in range(len(data)):
    time.append(data.iloc[i,0]) ## takes value from row i, column 0 (time) and appends it to list 'time'
    conc.append(data.iloc[i,1]) ## takes value from row i, column 1 (concentration) and appends it to list 'conc'
    
print("Data has been read!")
```

```python
# Question 5a

# Compute rate of change (conc_prime)
conc_prime = []

for i in range(len(time)):
    if i == len(time) - 1:  # Backward difference for the last point
        rate = (conc[i] - conc[i - 1]) / (time[i] - time[i - 1])
    else:  # Forward difference for all other points
        rate = (conc[i + 1] - conc[i]) / (time[i + 1] - time[i])
    conc_prime.append(rate)

print("Rate of change (conc_prime) computed!")
# test case
print(conc_prime[234]) ## should print 0.026448000000005203
```

```python
import matplotlib.pyplot as plt

max_rate = max(conc_prime)
max_time = time[conc_prime.index(max_rate)]

# Print the results
print(f"Maximum rate of change: {max_rate}")
print(f"Time at maximum rate: {max_time}")

# Plot the data
plt.figure(figsize=(10, 6))
plt.plot(time, conc_prime, label="Rate of Change of Concentration")
plt.scatter([max_time], [max_rate], color="red", label=f"Max Rate at t={max_time}")
plt.xlabel("Time (hours)")
plt.ylabel("Rate of Change of Concentration (mg/mL per hour)")
plt.title("Rate of Change of Concentration vs Time")
plt.legend()
plt.grid()
plt.show()
```

```python
# Question 7 

import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

def f(x):
    mu = 1     # Mean
    sigma = 2  # Standard deviation
    return (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma)**2)
        
a=-6
b=8
n=500
x=np.linspace(a,b,n+1)

plt.title("Graph of f(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.plot(x,f(x),color='blue') 
plt.grid()
plt.show()
```

```python
# 7a midpoint sum

def midpoint_sum(f,a,b,n):
    h = (b - a) / n  # Width of each subinterval
    midpoints = [a + h * (i + 0.5) for i in range(n)]  # Midpoints of each subinterval
    return sum(f(x) for x in midpoints) * h
    
    
    
print(midpoint_sum(f,1,3,100))  # should be 0.34134272963911727  
                                # possibly different in the last few decimals
    
# 7b trapezoidal rule

def trapezoidal_rule(f,a,b,n):
    h = (b - a) / n  # Width of each subinterval
    x_points = [a + i * h for i in range(n + 1)]  # Points dividing the intervals
    return h * (0.5 * f(a) + 0.5 * f(b) + sum(f(x) for x in x_points[1:-1]))
    
    
    
print(trapezoidal_rule(f,1,3,100))  # should be 0.3413427296391172 
                                # possibly different in the last few decimals
```
